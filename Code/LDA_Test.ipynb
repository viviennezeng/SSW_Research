- Count Vectorizer: Turn text into numbers (for chi square and LDA)
- TF-IDF (Term Frequency × Inverse Document Frequency):
  - Frequent terms that appear in many docs (like “the”, “file”, “data”) get low weight
  - Rarer but distinctive terms get high weight.
  1. compute TF-IDF vectors for every PR
  2. extract each PR’s mean IDF (“uncommonness”), higher IDF → rarer vocabulary
- Chi Square: Identify which words are most strongly associated with “scientific” (SK = 1) vs “non-scientific” (SK = 0) PRs.
  1. chi2(X_counts, y) computes χ² and p-values for all terms.
    - Top positive terms → appear more in scientific PRs.
    - Top negative terms → appear more in non-scientific PRs.
- LDA:
  - Positive Δ → topic appears more often in scientific PRs.
  - Negative Δ → topic appears more often in non-scientific PRs.




```python
# Sci vs Non-Sci PRs: Length, Uncommonness, Domain Vocabulary + LDA
import os, re, math, json
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.feature_selection import chi2
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Pretty display
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 120)
pd.set_option("display.precision", 4)

# Config
SCI_PATH = "dataset_scientific.csv"
NON_PATH = "dataset_non_scientific.csv"
LABEL_COL = "SK"          # 1 = scientific, 0 = non-scientific
N_TOPICS  = 12
N_TOP_WORDS = 12
TOP_K_DOMAIN_TERMS = 500
MIN_DF = 5        # ignore extremely rare terms
RANDOM_STATE = 42

# Load
def load_one(path, label):
    df = pd.read_csv(path)
    df[LABEL_COL] = label
    return df

sci = load_one(SCI_PATH, 1)
non = load_one(NON_PATH, 0)
df = pd.concat([sci, non], ignore_index=True)

# Build unified & fielded text
CAND_TITLE = ["title", "pr_title", "issue_title", "summary"]
CAND_DESC  = ["description", "body", "pr_body", "issue_body", "text", "body_text", "combined_text",
              "discussion", "comments", "comment_text"]

def pick_first(cols, candidates):
    for c in candidates:
        if c in cols:
            return c
    return None

def clean(s: str) -> str:
    s = str(s)
    s = s.lower()
    s = re.sub(r"[\r\n]+", " ", s)
    s = re.sub(r"https?://\S+|www\.\S+", " ", s)
    s = re.sub(r"[^\w\s\.\-\+/#]", " ", s)  # keep some code-y symbols
    s = re.sub(r"\s+", " ", s).strip()
    return s

cols = set(df.columns)
title_col = pick_first(cols, CAND_TITLE)
desc_col  = pick_first(cols, CAND_DESC)

if title_col is None:
    title_col = "title_fallback"
    df[title_col] = ""

if desc_col is None:
    # fall back to concatenating all object columns
    obj_cols = [c for c in df.columns if df[c].dtype == "object"]
    df["__fallback_text__"] = df[obj_cols].astype(str).agg(" ".join, axis=1)
    desc_col = "__fallback_text__"

df["title_clean"] = df[title_col].astype(str).map(clean)
df["desc_clean"]  = df[desc_col].astype(str).map(clean)

# Drop empty rows
df["__concat__"] = (df["title_clean"].fillna("") + " " + df["desc_clean"].fillna("")).str.strip()
df = df[df["__concat__"].str.len() > 0].reset_index(drop=True)

# Basic length & ratios
df["title_chars"] = df["title_clean"].str.len()
df["title_words"] = df["title_clean"].str.split().map(len)
df["desc_chars"]  = df["desc_clean"].str.len()
df["desc_words"]  = df["desc_clean"].str.split().map(len)

df["digit_ratio"]  = df["desc_clean"].str.count(r"\d").astype(float) / df["desc_chars"].replace(0, np.nan)
df["symbol_ratio"] = df["desc_clean"].str.count(r"[\.\-\+/#]").astype(float) / df["desc_chars"].replace(0, np.nan)

# Use CountVectorizer for LDA / chi2
count_vect = CountVectorizer(
    lowercase=True,
    # Remove common English words like "the," "is," and "a".
    stop_words="english",
    # Start with a letter (not a digit or punctuation), followed by alphanumerics or hyphens.
    token_pattern=r"(?u)\b[^\d\W][\w+\-]+\b",
    ngram_range=(1,2),
    # Ignores words that appear in fewer than MIN_DF PRs.
    min_df=MIN_DF
)
X_counts = count_vect.fit_transform(df["__concat__"])
vocab = np.array(count_vect.get_feature_names_out())

# Use TF-IDF for "uncommonness"
tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    token_pattern=r"(?u)\b[^\d\W][\w+\-]+\b",
    ngram_range=(1,2),
    min_df=MIN_DF
)
X_tfidf_all = tfidf.fit_transform(df["__concat__"])
tfidf_vocab = np.array(tfidf.get_feature_names_out())
idf_vals = tfidf.idf_

# Compute mean (weighted) IDF per field using same global IDF
def mean_weighted_idf(X, idf_array, row_idx):
    start, end = X.indptr[row_idx], X.indptr[row_idx+1]
    idxs = X.indices[start:end]
    dat  = X.data[start:end]
    if end - start == 0:
        return np.nan
    return np.average(idf_array[idxs], weights=dat)

X_title_tfidf = tfidf.transform(df["title_clean"])
X_desc_tfidf  = tfidf.transform(df["desc_clean"])

df["title_mean_idf"] = [mean_weighted_idf(X_title_tfidf, idf_vals, i) for i in range(X_title_tfidf.shape[0])]
df["desc_mean_idf"]  = [mean_weighted_idf(X_desc_tfidf,  idf_vals, i) for i in range(X_desc_tfidf.shape[0])]

# Characteristic via chi2
# Using chi2 square measures how strongly the presence of a term is associated with a particular class label.
# The p-value indicates the statistical significance of chi2 value.
y = df[LABEL_COL].values
chi2_scores, pvals = chi2(X_counts, y)
term_df = pd.DataFrame({
    "term": vocab,
    "chi2": chi2_scores,
    "pval": pvals
}).sort_values("chi2", ascending=False)

X_csr = X_counts.tocsr()
sk1 = (y == 1)
sk0 = ~sk1
sk1_means = np.asarray(X_csr[sk1].mean(axis=0)).ravel()
sk0_means = np.asarray(X_csr[sk0].mean(axis=0)).ravel()
label = np.sign(sk1_means - sk0_means)
term_df["label"] = np.where(label > 0, "SK=1", "SK=0")

top_sci_terms    = term_df[term_df["label"]=="SK=1"].head(40).copy()
top_nonsci_terms = term_df[term_df["label"]=="SK=0"].head(40).copy()

# Domain vocabulary & fractions (Title vs Description)
sci_terms = set(term_df[term_df["label"]=="SK=1"]["term"].head(TOP_K_DOMAIN_TERMS).tolist())
v2i = {t:i for i,t in enumerate(vocab)}
sci_idx = np.array([v2i[t] for t in sci_terms if t in v2i])

X_title_counts = count_vect.transform(df["title_clean"])
X_desc_counts  = count_vect.transform(df["desc_clean"])

def frac_in_domain_vocabulary(X_csr, lex_idx):
    indptr, indices = X_csr.indptr, X_csr.indices
    out = np.zeros(X_csr.shape[0], dtype=float)
    for i in range(X_csr.shape[0]):
        start, end = indptr[i], indptr[i+1]
        cols = indices[start:end]
        if end - start == 0:
            out[i] = 0.0
        else:
            hits = np.intersect1d(cols, lex_idx, assume_unique=False).size
            out[i] = hits / (end - start)
    return out

df["title_sci_vocab_frac"] = frac_in_domain_vocabulary(X_title_counts, sci_idx)
df["desc_sci_vocab_frac"]  = frac_in_domain_vocabulary(X_desc_counts,  sci_idx)

# LDA topics
lda = LatentDirichletAllocation(
    n_components=N_TOPICS,
    learning_method="batch",
    random_state=RANDOM_STATE,
    max_iter=20
)
lda.fit(X_counts)
doc_topic = lda.transform(X_counts)
topic_cols = [f"topic_{i:02d}" for i in range(N_TOPICS)]
doc_topic_df = pd.DataFrame(doc_topic, columns=topic_cols)

def topic_top_words(k, topn=N_TOP_WORDS):
    comp = lda.components_[k]
    idx = np.argsort(comp)[::-1][:topn]
    return [vocab[i] for i in idx]

topic_wordlists = {f"topic_{i:02d}": topic_top_words(i) for i in range(N_TOPICS)}

# Topic prevalence by class
topic_means = pd.concat([pd.Series(y, name=LABEL_COL), doc_topic_df], axis=1)\
                .groupby(LABEL_COL)[topic_cols].mean().T
topic_means.columns = ["non_sci_mean","sci_mean"] # columns: 0,1 ordered → rename for clarity
topic_means["delta_sci_minus_non_sci"] = topic_means["sci_mean"] - topic_means["non_sci_mean"]
topic_means["top_words"] = [", ".join(topic_wordlists[c][:8]) for c in topic_means.index]

# Stats helpers
def welch_p(a, b):
    a = pd.Series(a).dropna()
    b = pd.Series(b).dropna()
    if len(a) < 3 or len(b) < 3: return np.nan
    return stats.ttest_ind(a, b, equal_var=False).pvalue

def cohens_d(a, b):
    a = pd.Series(a).dropna().astype(float)
    b = pd.Series(b).dropna().astype(float)
    if len(a) < 3 or len(b) < 3: return np.nan
    na, nb = len(a), len(b)
    va, vb = a.var(ddof=1), b.var(ddof=1)
    s = np.sqrt(((na-1)*va + (nb-1)*vb) / (na+nb-2))
    if s == 0: return 0.0
    return (a.mean() - b.mean()) / s

def sig_stars(p):
    if pd.isna(p): return ""
    if p < 1e-6: return "****"
    if p < 1e-4: return "***"
    if p < 1e-2: return "**"
    if p < 5e-2: return "*"
    return ""

# Title vs Description Summary
metrics = [
    ("title_chars",         "Title length (chars)"),
    ("title_words",         "Title length (words)"),
    ("title_mean_idf",      "Title uncommonness (mean IDF)"),
    ("title_sci_vocab_frac","Title domain-vocabulary fraction"),
    ("desc_chars",          "Description length (chars)"),
    ("desc_words",          "Description length (words)"),
    ("desc_mean_idf",       "Description uncommonness (mean IDF)"),
    ("desc_sci_vocab_frac", "Description domain-vocabulary fraction"),
]

rows = []
for col, nice in metrics:
    a = df.loc[df[LABEL_COL]==1, col]  # sci
    b = df.loc[df[LABEL_COL]==0, col]  # non-sci
    sci_mean = a.mean()
    nonsci_mean = b.mean()
    delta = sci_mean - nonsci_mean
    p = welch_p(a, b)
    d = cohens_d(a, b)
    rows.append({
        "Metric": nice,
        "Mean (SK=1)": sci_mean,
        "Mean (SK=0)": nonsci_mean,
        "Δ (SK1 - SK0)": delta,
        "Cohen's d": d,
        "Welch p": p,
        "Sig": sig_stars(p)
    })

summary = pd.DataFrame(rows).sort_values("Δ (SK1 - SK0)", ascending=False)
summary_rounded = summary.copy()
for c in ["Mean (SK=1)", "Mean (SK=0)", "Δ (SK1 - SK0)", "Cohen's d", "Welch p"]:
    summary_rounded[c] = summary_rounded[c].map(lambda x: np.nan if pd.isna(x) else float(f"{x:.4g}"))

# Outputs
print("(A) Title & Description Comparisons (means, Δ, effect size, p)")
display(summary_rounded)

print("\nInterpretation tips:")
print("Positive Δ means SK=1 > SK=0. Sig stars: * p<0.05, ** p<0.01, *** p<1e-4, **** p<1e-6.")
print("Cohen’s d effect sizes: ~0.2 small, ~0.5 medium, ~0.8 large.")

print("(B) Characteristic Terms (chi²)")
print("\nTop SK=1 terms:")
display(top_sci_terms.reset_index(drop=True))
print("\nTop SK=0 terms:")
display(top_nonsci_terms.reset_index(drop=True))

print("(C) LDA Topics: class means & top words")
print("\nTopics that are more prevalent in scientific PRs:")
display(topic_means.sort_values("delta_sci_minus_non_sci", ascending=False).head(8))
print("\nTopics that are more prevalent in non scientific PRs:")
display(topic_means.sort_values("delta_sci_minus_non_sci", ascending=True).head(8))

# Charts
out_dir = Path("lda_outputs"); out_dir.mkdir(exist_ok=True)

def bar_compare(metric_key, nice_name):
    means = df.groupby(LABEL_COL)[metric_key].mean()
    fig = plt.figure(figsize=(5,4))
    plt.bar(["Non-Sci (0)", "Sci (1)"], [means.get(0,0), means.get(1,0)])
    plt.title(nice_name)
    plt.ylabel("Mean")
    plt.tight_layout()
    plt.show()
    fig.savefig(out_dir / f"mean_{metric_key}.png", dpi=160)

print("(D)) Visualization (means by class)")
for key, nice in [
    ("title_words", "Title length (words)"),
    ("desc_words", "Description length (words)"),
    ("title_mean_idf", "Title uncommonness (mean IDF)"),
    ("desc_mean_idf", "Description uncommonness (mean IDF)"),
    ("title_sci_vocab_frac", "Title domain-vocabulary fraction"),
    ("desc_sci_vocab_frac", "Description domain-vocabulary fraction")
]:
    bar_compare(key, nice)

# Topic deltas bar chart
tm_sorted = topic_means.sort_values("delta_sci_minus_non_sci", ascending=False)
fig = plt.figure(figsize=(8,5))
plt.bar(tm_sorted.index, tm_sorted["delta_sci_minus_non_sci"].values)
plt.xticks(rotation=60)
plt.ylabel("Δ Topic Prevalence (Sci - Non)")
plt.title("Topics more prevalent in Sci (positive) vs Non-Sci (negative)")
plt.tight_layout()
plt.show()
fig.savefig(out_dir / "topics_delta_bar.png", dpi=160)

# Save
summary.to_csv(out_dir / "sci_vs_non_sci_title_desc_summary.csv", index=False)
term_df.to_csv(out_dir / "characteristic_term_chi2.csv", index=False)
pd.DataFrame({
    "topic": list(topic_wordlists.keys()),
    "top_words": [", ".join(words) for words in topic_wordlists.values()]
}).to_csv(out_dir / "lda_topics_top_words.csv", index=False)
topic_means.to_csv(out_dir / "topic_means_by_class.csv")
doc_export_cols = [
    LABEL_COL,
    "title_chars","title_words","title_mean_idf","title_sci_vocab_frac",
    "desc_chars","desc_words","desc_mean_idf","desc_sci_vocab_frac"
] + topic_cols
pd.concat([df[doc_export_cols[:9]], doc_topic_df], axis=1).to_csv(out_dir / "doc_level_features_and_topics.csv", index=False)

print("\nSaved CSVs & PNGs in:", out_dir.resolve())

# Auto-generated bullets
def arrow(x):
    return "↑" if x > 0 else ("↓" if x < 0 else "≈")

bullets = []
for _, r in summary.iterrows():
    bullets.append(
        f"- {r['Metric']}: Δ={r['Δ (SK1 - SK0)']:.3f} {arrow(r['Δ (SK1 - SK0)'])}, "
        f"d={r['Cohen\'s d']:.2f}, p={r['Welch p']:.2e}{sig_stars(r['Welch p'])}"
    )

print("\nSuggested bullets (paste & edit in manuscript):\n")
print("\n".join(bullets))
```
